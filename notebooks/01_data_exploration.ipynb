## Cell 2: load the sentiment data and count classes

```python
from datasets import load_dataset
import pandas as pd

sent = load_dataset("cardiffnlp/tweet_eval", "sentiment")
print(sent)

label_names = {0: "negative", 1: "neutral", 2: "positive"}
splits = {
    "train": sent["train"].to_pandas(),
    "validation": sent["validation"].to_pandas(),
    "test": sent["test"].to_pandas(),
}

summary = pd.DataFrame(
    {name: df["label"].map(label_names).value_counts() for name, df in splits.items()}
)
print(summary)
print((summary / summary.sum() * 100).round(1))
```

## Cell 3: read 30 examples by eye

```python
train = splits["train"]
for _, row in train.sample(30, random_state=42).iterrows():
    print(f"[{label_names[row['label']]}] {row['text']}\n")
```

## Cell 4: the irony set, for your hard-set work later

```python
irony = load_dataset("cardiffnlp/tweet_eval", "irony")
irony_train = irony["train"].to_pandas()
irony_names = {0: "not ironic", 1: "ironic"}
print(irony_train["label"].map(irony_names).value_counts())

for _, row in irony_train[irony_train["label"] == 1].sample(10, random_state=42).iterrows():
    print(row["text"], "\n")
```
